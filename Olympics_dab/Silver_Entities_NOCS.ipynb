## Silver Nootbook

## Reading NOCS Data

```python
df = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("abfss://bronze@saolympicsproj.dfs.core.windows.net/nocs")
```

```python
display(df)
```

## Droping Columns

```python
df = df.drop('country')
```

```python
df.display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = df.withColumn('tag',split(col('tag'),'-')[0])
df.display()
```

```python
df.write.format("delta")\
    .mode("append")\
    .option("path","abfss://silver@saolympicsproj.dfs.core.windows.net/nocs")\
    .saveAsTable("olympics.silver.nocs")
```
